## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# 1. Clean df1

# Fill missing values in the 'character' column using the most frequent value (mode)
df1['character'].fillna(df1['character'].mode()[0], inplace=True)

# Remove duplicate rows to ensure data consistency
df1.drop_duplicates(inplace=True)


# 2. Clean df2

# Drop unnecessary columns 'seasons' and 'age_certification' if they exist
df2.drop(['seasons', 'age_certification'], axis=1, inplace=True, errors='ignore')

# Replace missing values in 'description' column with 'unknown'
df2['description'].fillna('unknown', inplace=True)

# Replace missing values in 'imdb_id' column with 'unknown'
df2['imdb_id'].fillna('unknown', inplace=True)

# List of numerical columns for handling missing values
num_cols = ['tmdb_score','tmdb_popularity','imdb_score','imdb_votes']

# Fill missing values in numerical columns using the median of each column
for col in num_cols:
    df2[col].fillna(df2[col].median(), inplace=True)

# Remove duplicate rows
df2.drop_duplicates(inplace=True)


# 3. Merge (AFTER cleaning)

# Merge df1 and df2 on the 'id' column using inner join
df = pd.merge(df2, df1, on='id', how='inner')


# 4. Final checks

# Standardize column names: convert to lowercase and replace spaces with underscores
df.columns = df.columns.str.lower().str.replace(" ", "_")

# Display dataset information (data types, non-null counts)
df.info()

# Display statistical summary of numerical columns
df.describe()

# Check remaining missing values in each column
df.isnull().sum()
````

### What all manipulations have you done and insights you found?

Data Wrangling and Manipulations

In this project, several data wrangling steps were performed to clean and prepare the dataset for analysis:

1. Handling Missing Values:  
 Missing values in the character column were filled using the mode (most frequent value). Numerical columns such as imdb_score, imdb_votes, tmdb_score, and tmdb_popularity were filled using the median to reduce the impact of outliers. Text-based columns like description and imdb_id were filled with the value “unknown”.

2. Dropping Irrelevant Columns:  
Columns such as seasons and age_certification were removed as they were not relevant for the analysis.

3. Removing Duplicates:
 Duplicate records were removed from both datasets to ensure data accuracy and avoid redundancy.

4. Merging Datasets:
 The two datasets were merged using an inner join on a common column (id) to retain only consistent and matching records.

5. Column Standardization:
  Column names were converted to lowercase and spaces were replaced with underscores to maintain consistency and improve readability.
